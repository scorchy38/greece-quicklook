# Island populations

A quick look at how Greece's islands compare. Density is people per km², $\rho = P / A$.

```python
import pandas as pd

islands = pd.read_csv("Islands.csv")
islands["density"] = (islands["Population"] / islands["Area (km²)"]).round(1)
```

```python
cyclades = islands.query("Group == 'Cyclades'")
cyclades.nlargest(4, "density")[["Island", "Population", "Area (km²)", "density"]]
```

## The largest islands

Crete alone has more people than the next five islands combined.

```python
top = islands.nlargest(6, "Population")
print(f"Plotting {len(top)} islands…")
top.plot.barh(x="Island", y="Population", legend=False)
```
